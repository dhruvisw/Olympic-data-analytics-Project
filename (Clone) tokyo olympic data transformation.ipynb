```python
# Load transformed data
athletes_transformed = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/dhruvi_data/transformed_data/athletes")
coaches_transformed = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/dhruvi_data/transformed_data/coaches")
entriesgender_transformed = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/dhruvi_data/transformed_data/entriesgender")
medals_transformed = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/dhruvi_data/transformed_data/medals")
teams_transformed = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/dhruvi_data/transformed_data/teams")
```

```python
%python
from pyspark.sql.functions import sum

# Total medals by team country
total_medals_by_team = medals_transformed.groupBy("TeamCountry").agg(
    sum(medals_transformed["Gold"]).alias("Total Gold"),
    sum(medals_transformed["Silver"]).alias("Total Silver"),
    sum(medals_transformed["Bronze"]).alias("Total Bronze")
)

# Display as a stacked bar chart
display(total_medals_by_team)
```

```python
%python
# Calculate male/female ratio by discipline
gender_distribution_by_sport = entriesgender_transformed.groupBy("Discipline").agg(
    sum("Male").alias("Total Male"),
    sum("Female").alias("Total Female")
)

# Display as pie chart
display(gender_distribution_by_sport)
```

```python
%python
# Gold medals by country and rank
gold_medals_by_country_rank = medals_transformed.groupBy("TeamCountry", "Rank").agg(sum("Gold").alias("Gold"))

# Display as a heatmap or bar chart
display(gold_medals_by_country_rank)
```

```python
%python
from pyspark.sql.functions import col

# Example use of `col`
coach_count = coaches_transformed.groupBy("Country").count().withColumnRenamed("count", "CoachCount")
athlete_count = athletes_transformed.groupBy("Country").count().withColumnRenamed("count", "AthleteCount")

coach_athlete_ratio = coach_count.join(athlete_count, "Country") \
    .withColumn("CoachAthleteRatio", col("CoachCount") / col("AthleteCount"))

display(coach_athlete_ratio)
```

```python
%python
athletes_by_country = athletes_transformed.groupBy("Country").count().orderBy("count", ascending=False)
display(athletes_by_country)
```

```python
%python
top_medal_countries = medals_transformed.withColumn("Total", col("Gold") + col("Silver") + col("Bronze")) \
    .orderBy("Total", ascending=False) \
    .select("TeamCountry", "Gold", "Silver", "Bronze", "Total") \
    .limit(10)

display(top_medal_countries)

```
